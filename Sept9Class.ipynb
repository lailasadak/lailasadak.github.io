```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0, 10, 100)

# Two functions
y1 = x**2
y2 = 2*x + 1

# Create the plot
fig, ax = plt.subplots()

# Plot each function in a different color
ax.plot(x, y1, color="blue", label="y = x²")
ax.plot(x, y2, color="red", label="y = 2x + 1")

# Labels
ax.set_xlabel("x-axis")
ax.set_ylabel("y-axis")

# Show legend
ax.legend()

# Show grid
ax.grid(True)

plt.show()
```
