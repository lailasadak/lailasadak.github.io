```python
import pandas as pd
import glob

# Find the Excel file in Google Colab
files = glob.glob('/content/*.xlsx')

print("Excel files found:", files)

# Load the first Excel file found
df = pd.read_excel(files[0])

# Show the column names so we can verify them
print("Column names:", df.columns.tolist())

# Add up all base pairs
total_bp = df['basepais(bp)'].sum()

# Each base pair = 3.4 angstroms
total_angstroms = total_bp * 3.4

# Convert angstroms to meters
total_meters = total_angstroms * 1e-10

# Display results
print("Total base pairs:", total_bp)
print("Total length in angstroms:", total_angstroms)
print("Total length in meters:", total_meters)
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import glob

# Find the Excel file automatically
file = glob.glob('/content/*.xlsx')[0]

# Load the file
df = pd.read_excel(file)

# Find total base pairs
total_bp = df['basepais(bp)'].sum()

# Years and cost per million base pairs
years = [2001, 2010, 2011, 2020]
cost_per_Mbp = [10000, 1, 0.10, 0.01]

# Calculate total sequencing cost
total_cost = [(total_bp / 1_000_000) * cost for cost in cost_per_Mbp]

# Print the results
for year, cost in zip(years, total_cost):
    print(f"{year}: ${cost:,.2f}")

# Make the plot
plt.plot(years, total_cost, marker='o')
plt.xlabel("Year")
plt.ylabel("Total Sequencing Cost ($)")
plt.title("Cost to Sequence the Whole Genome")
plt.grid(True)
plt.show()
```
