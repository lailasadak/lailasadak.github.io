```python
from sklearn.manifold import TSNE
from sklearn.preprocessing import StandardScaler

integrins
```

```python
X = integrins.select_dtypes(include='number')

labels = integrins['primary_site']
```

```python
scaler = StandardScaler()

X_scaled = scaler.fit_transform(X)
```

```python
tsne = TSNE(
    n_components=2,
    random_state=42
)

X_2d = tsne.fit_transform(X_scaled)
```

```python
plt.figure(figsize=(10, 6))

sns.scatterplot(
    x=X_2d[:, 0],
    y=X_2d[:, 1],
    hue=labels,
    palette='tab10',
    s=30
)

plt.title("Integrin Gene Expression Levels")
plt.xlabel("t-SNE 1")
plt.ylabel("t-SNE 2")

plt.legend(
    title="Primary Site",
    bbox_to_anchor=(1.05, 1),
    loc="upper left"
)

plt.tight_layout()
plt.show()
```
